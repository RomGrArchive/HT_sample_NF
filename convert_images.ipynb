```python
from PIL import Image
import os
import shutil

input_dir = './images'
output_dir = './images_webp'
os.makedirs(output_dir, exist_ok=True)

quality = 80

for filename in os.listdir(input_dir):
    filepath = os.path.join(input_dir, filename)

    if filename.lower().endswith(('.png', '.jpg', '.jpeg')):
        base_name = os.path.splitext(filename)[0]
        try:
            with Image.open(filepath) as img:
                # Resize image to half its original dimensions
                new_size = (img.width // 2, img.height // 2)
                img_resized = img.resize(new_size, Image.LANCZOS)

                webp_path = os.path.join(output_dir, f"{base_name}.webp")
                img_resized.save(webp_path, 'WEBP', quality=quality)
                print(f"Saved image: {webp_path}")
        except Exception as e:
            print(f"Error with image {filename}: {e}")
    else:
        # Copy non-image file directly
        dest_path = os.path.join(output_dir, filename)
        try:
            shutil.copy2(filepath, dest_path)
            print(f"Copied file: {dest_path}")
        except Exception as e:
            print(f"Error copying {filename}: {e}")
```
